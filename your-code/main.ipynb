In this lab, we will explore a dataset that describes websites with different features and labels them either benign or malicious . We will use supervised learning algorithms to figure out what feature patterns malicious websites are likely to have and use our model to predict malicious websites.

Your features will be:

+ URL: it is the anonymous identification of the URL analyzed in the study
+ URL_LENGTH: it is the number of characters in the URL
+ NUMBER_SPECIAL_CHARACTERS: it is number of special characters identified in the URL, such as, “/”, “%”, “#”, “&”, “. “, “=”
+ CHARSET: it is a categorical value and its meaning is the character encoding standard (also called character set).
+ SERVER: it is a categorical value and its meaning is the operative system of the server got from the packet response.
+ CONTENT_LENGTH: it represents the content size of the HTTP header.
+ WHOIS_COUNTRY: it is a categorical variable, its values are the countries we got from the server response (specifically, our script used the API of Whois).
+ WHOIS_STATEPRO: it is a categorical variable, its values are the states we got from the server response (specifically, our script used the API of Whois).
+ WHOIS_REGDATE: Whois provides the server registration date, so, this variable has date values with format DD/MM/YYY HH:MM
+ WHOIS_UPDATED_DATE: Through the Whois we got the last update date from the server analyzed
+ TCP_CONVERSATION_EXCHANGE: This variable is the number of TCP packets exchanged between the server and our honeypot client
+ DIST_REMOTE_TCP_PORT: it is the number of the ports detected and different to TCP
+ REMOTE_IPS: this variable has the total number of IPs connected to the honeypot
+ APP_BYTES: this is the number of bytes transfered
+ SOURCE_APP_PACKETS: packets sent from the honeypot to the server
+ REMOTE_APP_PACKETS: packets received from the server
+ APP_PACKETS: this is the total number of IP packets generated during the communication between the honeypot and the server
+ DNS_QUERY_TIMES: this is the number of DNS packets generated during the communication between the honeypot and the server
+ TYPE: this is a categorical variable, its values represent the type of web page analyzed, specifically, 1 is for malicious websites and 0 is for benign websites

# Challenge 3 - Handle Missing Values

The next step would be handling missing values. **We start by examining the number of missing values in each column, which you will do in the next cell.**

````python
# Your code here
missing = websites.isnull().sum()
pct = 100 * websites.isnull().mean()
report = pd.DataFrame({"missing": missing, "percent": pct.round(1)})
print(report[report.missing > 0].sort_values("missing", ascending=False).to_string())
````

````python
# Firts we will drop the columns with more than 50% of missing data
threshold = 0.5
over = websites.columns[websites.isnull().mean() > threshold]
print("columns over 50%% missing:", list(over) if len(over) else "none")

# Nothing crosses the 50% line. But CONTENT_LENGTH is missing in 45.6% of
# rows, just under it - and the next step drops every row with any missing
# value. Keeping CONTENT_LENGTH therefore costs most of the dataset:
a = len(websites.dropna())
b = len(websites.drop(columns=["CONTENT_LENGTH"]).dropna())
print()
print("rows surviving dropna() keeping CONTENT_LENGTH : %d" % a)
print("rows surviving dropna() dropping CONTENT_LENGTH: %d" % b)
print("-> dropping one column saves %d rows" % (b - a))

websites = websites.drop(columns=over.tolist() + ["CONTENT_LENGTH"])
print()
print("dropped CONTENT_LENGTH. columns now:", websites.shape[1])
````

````python
# Now we will drop the rows with missing data
before = len(websites)
before_mal = int((websites.Type == 1).sum())

websites = websites.dropna()

print("rows: %d -> %d  (%.1f%% lost)"
      % (before, len(websites), 100 * (1 - len(websites) / before)))
print("malicious: %d -> %d" % (before_mal, int((websites.Type == 1).sum())))
print()
print("class balance now:")
print(websites.Type.value_counts().to_string())

# Dropping rows is the blunt option. It is defensible here because the
# remaining missing values are in categorical columns where imputing a
# country or a server name would be inventing data. It does cost us a
# quarter of the rows, and the minority class shrinks with them.
````

Since we have fixed the country values, can we convert this column to ordinal now?

Not yet. If you reflect on the previous labs how we handle categorical columns, you probably remember we ended up dropping a lot of those columns because there are too many unique values. Too many unique values in a column is not desirable in machine learning because it makes prediction inaccurate. But there are workarounds under certain conditions. One of the fixable conditions is:

#### If a limited number of values account for the majority of data, we can retain these top values and re-label all other rare values.

The `WHOIS_COUNTRY` column happens to be this case. You can verify it by print a bar chart of the `value_counts` in the next cell to verify:

````python
def print_bar_plot(x,y):
    plt.bar(x, y)
    plt.figure(figsize=(15,15))
    return plt.show()
````

````python
print_bar_plot(websites.WHOIS_COUNTRY.unique(),websites.WHOIS_COUNTRY.value_counts());
````

#### After verifying, now let's keep the top 10 values of the column and re-label other columns with `OTHER`.

````python
# Your code here
# Keep the top 10 countries, re-label everything else as OTHER
top10 = websites.WHOIS_COUNTRY.value_counts().head(10).index.tolist()
print("top 10 countries:", top10)

websites.WHOIS_COUNTRY = websites.WHOIS_COUNTRY.apply(
    lambda x: x if x in top10 else "OTHER")

print()
print(websites.WHOIS_COUNTRY.value_counts().to_string())
print()
print("unique values now:", websites.WHOIS_COUNTRY.nunique())
````

Now since `WHOIS_COUNTRY` has been re-labelled, we don't need `WHOIS_STATEPRO` any more because the values of the states or provinces may not be relevant any more. We'll drop this column.

In addition, we will also drop `WHOIS_REGDATE` and `WHOIS_UPDATED_DATE`. These are the registration and update dates of the website domains. Not of our concerns.

#### In the next cell, drop `['WHOIS_STATEPRO', 'WHOIS_REGDATE', 'WHOIS_UPDATED_DATE']`.

````python
# Your code here
# WHOIS_STATEPRO is no longer meaningful once countries are grouped, and the
# two date columns are free-text dates with hundreds of unique values.
drop_cols = ["WHOIS_STATEPRO", "WHOIS_REGDATE", "WHOIS_UPDATED_DATE"]
websites = websites.drop(columns=drop_cols)
print("dropped:", drop_cols)
print("columns now:", list(websites.columns))
````

At the end, your `SERVER` column should only contain 4 unique values: `Microsoft`, `Apache`, `nginx`, and `Other`.

````python
# Count `SERVER` value counts here
# Collapse into the three main server families; anything else becomes Other.
def simplify_server(value):
    v = str(value).lower()
    if "apache" in v:
        return "Apache"
    if "nginx" in v:
        return "nginx"
    if "microsoft" in v or "iis" in v:
        return "Microsoft"
    return "Other"


websites.SERVER = websites.SERVER.apply(simplify_server)

print(websites.SERVER.value_counts().to_string())
print()
print("unique values: %d (was 238)" % websites.SERVER.nunique())

# "Other" is not a dustbin here - it keeps every server that is none of the
# three, which is a real signal rather than noise to be discarded.
````

# Bonus Challenge - Feature Scaling

Problem-solving in machine learning is iterative. You can improve your model prediction with various techniques (there is a sweetspot for the time you spend and the improvement you receive though). Now you've completed only one iteration of ML analysis. There are more iterations you can conduct to make improvements. In order to be able to do that, you will need deeper knowledge in statistics and master more data analysis techniques. In this bootcamp, we don't have time to achieve that advanced goal. But you will make constant efforts after the bootcamp to eventually get there.

However, now we do want you to learn one of the advanced techniques which is called *feature scaling*. The idea of feature scaling is to standardize/normalize the range of independent variables or features of the data. This can make the outliers more apparent so that you can remove them. This step needs to happen during Challenge 6 after you split the training and test data because you don't want to split the data again which makes it impossible to compare your results with and without feature scaling. For general concepts about feature scaling, click [here](https://en.wikipedia.org/wiki/Feature_scaling). To read deeper, click [here](https://medium.com/greyatom/why-how-and-when-to-scale-your-features-4b30ab09db5e).

In the next cell, attempt to improve your model prediction accuracy by means of feature scaling. A library you can utilize is `sklearn.preprocessing.RobustScaler` ([documentation](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.RobustScaler.html)). You'll use the `RobustScaler` to fit and transform your `X_train`, then transform `X_test`. You will use logistic regression to fit and predict your transformed data and obtain the accuracy score in the same way. Compare the accuracy score with your normalized data with the previous accuracy data. Is there an improvement?

````python
# Your code here
# Bonus: scale the features and refit both models.
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)   # fit on TRAIN only
X_test_scaled = scaler.transform(X_test)         # transform test with the same scaler

# Fitting the scaler on all the data before splitting would leak information
# from the test set into the training step.

print("%-34s %8s %8s %10s" % ("model", "acc", "recall", "precision"))
print("-" * 64)


def report(name, model, Xtr, Xte):
    model.fit(Xtr, y_train)
    pred = model.predict(Xte)
    cm = confusion_matrix(y_test, pred)
    tn, fp, fn, tp = cm.ravel()
    rec = tp / (tp + fn) if (tp + fn) else 0
    prec = tp / (tp + fp) if (tp + fp) else 0
    print("%-34s %8.4f %8.3f %10.3f"
          % (name, accuracy_score(y_test, pred), rec, prec))


report("logistic regression, unscaled", LogisticRegression(max_iter=5000, random_state=42),
       X_train, X_test)
report("logistic regression, scaled", LogisticRegression(max_iter=5000, random_state=42),
       X_train_scaled, X_test_scaled)
report("tree depth=5, unscaled", DecisionTreeClassifier(max_depth=5, random_state=42),
       X_train, X_test)
report("tree depth=5, scaled", DecisionTreeClassifier(max_depth=5, random_state=42),
       X_train_scaled, X_test_scaled)

print()
print("Scaling should move logistic regression and leave the tree unchanged:")
print("a decision tree splits on thresholds within one feature at a time, so")
print("a monotonic rescaling of that feature does not change which rows fall")
print("on which side of a split. Distance- and gradient-based models care.")
````

## Comments:

**Does scaling help?**

```
model                                   acc   recall  precision
logistic regression, unscaled        0.9563    0.643      0.947
logistic regression, scaled          0.9484    0.536      1.000
tree depth=5, unscaled               0.9405    0.571      0.842
tree depth=5, scaled                 0.9405    0.571      0.842
```

**The tree is completely unchanged**, exactly as expected. A decision tree splits on a threshold within one feature at a time, so rescaling that feature monotonically does not change which rows land on which side of a split. Scaling is irrelevant to trees.

**Logistic regression changed, but not for the better.** Accuracy fell slightly and recall dropped from 0.64 to 0.54, while precision rose to a perfect 1.000. The scaled model became *more* conservative: it now never raises a false alarm, at the cost of missing even more malicious sites. Given that false negatives are the expensive error here, that is a step backwards.

This is worth stating plainly because the expected answer is "scaling improves logistic regression". It did fix the convergence problem — the unscaled fit never converged even at 5000 iterations — so the scaled model is the better-behaved one numerically. But better-converged is not the same as better at the task.

**Other things that would plausibly help more than scaling:**

- `class_weight="balanced"` on either model, which addresses the 8:1 imbalance directly rather than indirectly
- Lowering the decision threshold below 0.5, trading precision for recall
- Cross-validated `max_depth` instead of the hand-picked 5 or unlimited
